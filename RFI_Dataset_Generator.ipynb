```python
from faker import Faker
import pandas as pd
import random
from datetime import timedelta
```

```python
fake = Faker()
```

```python
# Definitions
topics = [
    'Pavement Materials', 'Stormwater Drainage', 'Traffic Signage', 
    'Utility Relocation', 'Lane Closures', 'Guardrail Placement', 
    'Road Grading', 'Bridge Joint Design', 'Soil Compaction'
]
```

```python
statuses = ['Open', 'Answered', 'Closed', 'Pending Review']
reviewers = ['Civil Engineer', 'DOT Inspector', 'General Contractor', 'Subcontractor']
priority_levels = ['High', 'Medium', 'Low']
project_phases = ['Preconstruction', 'Excavation', 'Paving', 'Inspection']
rfi_types = ['Clarification', 'Design Change', 'Conflict Resolution', 'Material Substitution']
```

```python
data = []
```

```python
for i in range(1500):
    submission = fake.date_between(start_date='-180d', end_date='-14d')
    response_gap = random.randint(3, 30)
    response = submission + timedelta(days=response_gap)

    priority = random.choices(priority_levels, weights=[0.2, 0.5, 0.3])[0]  # more Medium
    project_phase = random.choice(project_phases)

    # Estimate cost impact based on priority
    if priority == 'High':
        cost_impact = round(random.uniform(50000, 200000), 2)
    elif priority == 'Medium':
        cost_impact = round(random.uniform(10000, 50000), 2)
    else:
        cost_impact = round(random.uniform(1000, 10000), 2)

    # Smarter delay logic based on meaningful features
    delay_prob = 0.2 if priority == 'Low' else 0.5 if priority == 'Medium' else 0.75
    if cost_impact > 100000:
        delay_prob += 0.1
    if project_phase in ['Paving', 'Inspection']:
        delay_prob += 0.1
    delay_prob = min(delay_prob, 0.95)

    delay = random.random() < delay_prob

    record = {
        'rfi_id': i + 1,
        'project_name': fake.city() + " Bypass Project",
        'submission_date': submission,
        'response_date': response,
        'status': random.choice(statuses),
        'topic': random.choice(topics),
        'description': fake.sentence(nb_words=18),
        'delay_flag': delay,
        'days_to_close': response_gap,
        'assigned_to': random.choice(reviewers),
        'priority_level': priority,
        'cost_impact_estimate': cost_impact,
        'project_phase': project_phase,
        'rfi_type': random.choice(rfi_types),
        'response_quality': random.randint(2, 5)
    }

    data.append(record)
```

```python
df = pd.DataFrame(data)
```
